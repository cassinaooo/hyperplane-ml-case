```python
import pickle as pkl

import numpy as np
import pandas as pd
from tqdm import tqdm

from normalize_vendor import normalize_vendor
```

#### Objective:

Using `sentence-transformers/all-mpnet-base-v2`, generates embeddings for all MCCs.

#### Depends on:

```python
fname = "./data/01-parsed-amounts-and-dates.parquet"
```

#### Generates:

```python
mcc_to_vector_fname = "./data/mcc_to_vector.pkl"
```

----------------

```python
df = pd.read_parquet(fname)
```

```python
df.head()
```

```python
mcc_strs = list(sorted(df.mcc.unique()))
```

```python
mcc_strs[:5]
```

```python
len(mcc_strs)
```

```python
norm_mcc_str = sorted(
    list(set(filter(lambda x: len(x) > 0, map(normalize_vendor, mcc_strs))))
)
```

```python
len(norm_mcc_str)
```

```python
norm_mcc_str[:10]
```

```python
from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer("sentence-transformers/all-mpnet-base-v2")
```

```python
mcc_to_vector = {}
```

```python
for i, norm_mcc in enumerate(norm_mcc_str):
    print(norm_mcc)

    embeddings = model.encode(norm_mcc)
    mcc_to_vector[norm_mcc] = {}
    mcc_to_vector[norm_mcc]["embedding"] = embeddings

print("done")
```

```python
with open(mcc_to_vector_fname, "wb") as f:
    pkl.dump(
        mcc_to_vector,
        f,
    )
```
